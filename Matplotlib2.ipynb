## Matplotlib plots (scatter, bar, histogram, pie chart)
### 11/03/26
    

```python
import matplotlib.pyplot as plt
```

### scatter plot

```python
# Education – Increasing trend
study_hours = [1,2,3,4,5]
marks = [35,50,65,80,92]

plt.scatter(study_hours, marks)
plt.title("Education: Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()


# Health – Decreasing trend
exercise_hours = [1,2,3,4,5]
body_fat = [30,27,24,20,17]

plt.scatter(exercise_hours, body_fat)
plt.title("Health: Exercise vs Body Fat %")
plt.xlabel("Exercise Hours")
plt.ylabel("Body Fat %")
plt.show()


# Business – Rapid growth
ads_budget = [100,200,300,400,500]
sales = [1200,1800,2600,3600,5000]

plt.scatter(ads_budget, sales)
plt.title("Business: Ads Budget vs Sales")
plt.xlabel("Ads Budget")
plt.ylabel("Sales Revenue")
plt.show()


# Agriculture – Peak pattern
rainfall = [50,100,150,200,250]
crop_yield = [20,45,70,48,25]

plt.scatter(rainfall, crop_yield)
plt.title("Agriculture: Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Crop Yield")
plt.show()


# Technology – Zigzag trend
experience = [1,2,3,4,5]
productivity = [40,70,55,85,65]

plt.scatter(experience, productivity)
plt.title("Technology: Experience vs Productivity")
plt.xlabel("Experience (Years)")
plt.ylabel("Productivity Score")
plt.show()
```

### Bar Chart

```python
# Product Sales
products = ["A","B","C","D","E"]
sales = [120,200,280,360,450]

plt.bar(products, sales)
plt.title("Business: Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()


# Course Enrollments
courses = ["Python","Java","SQL","PowerBI","Excel"]
students = [150,90,110,70,130]

plt.bar(courses, students)
plt.title("Education: Course Enrollments")
plt.xlabel("Courses")
plt.ylabel("Students")
plt.show()


# Olympic Medals
countries = ["USA","China","Japan","UK","India"]
medals = [120,95,70,65,25]

plt.bar(countries, medals)
plt.title("Sports: Olympic Medal Count")
plt.xlabel("Countries")
plt.ylabel("Medals")
plt.show()


# Monthly Rainfall
months = ["Jan","Feb","Mar","Apr","May"]
rainfall = [15,25,40,60,85]

plt.bar(months, rainfall)
plt.title("Weather: Monthly Rainfall")
plt.xlabel("Months")
plt.ylabel("Rainfall (mm)")
plt.show()


# Mobile Sales
brands = ["Apple","Samsung","Xiaomi","Oppo","Vivo"]
sales = [45,70,95,68,50]

plt.bar(brands, sales)
plt.title("Technology: Mobile Sales")
plt.xlabel("Brands")
plt.ylabel("Units Sold")
plt.show()
```

### Histogram

```python
# Student Marks
marks = [45,48,50,52,55,58,60,62,65]

plt.hist(marks)
plt.title("Education: Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()


# Salary Distribution
salaries = [25000,28000,30000,32000,35000,60000,90000]

plt.hist(salaries)
plt.title("Business: Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()


# Temperature Variation
temperatures = [20,23,26,29,32,35,38]

plt.hist(temperatures)
plt.title("Weather: Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.show()


# Internet Usage Hours
internet_hours = [1,2,2,3,3,3,4,4,5]

plt.hist(internet_hours)
plt.title("Technology: Internet Usage Hours")
plt.xlabel("Hours")
plt.ylabel("Frequency")
plt.show()


# Age Distribution
ages = [18,22,25,30,35,40,50,60]

plt.hist(ages)
plt.title("Population: Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### Pie Chart

```python
# Programming Language Usage
languages = ["Python","Java","JavaScript","C++","Go"]
usage = [40,25,20,10,5]

plt.pie(usage, labels=languages, autopct="%1.1f%%")
plt.title("Technology: Programming Language Usage")
plt.show()


# Monthly Budget
categories = ["Rent","Food","Transport","Savings","Entertainment"]
budget = [40,25,10,15,10]

plt.pie(budget, labels=categories, autopct="%1.1f%%")
plt.title("Finance: Monthly Budget Distribution")
plt.show()


# Energy Sources
energy = ["Coal","Solar","Wind","Hydro","Gas"]
share = [45,20,15,10,10]

plt.pie(share, labels=energy, autopct="%1.1f%%")
plt.title("Energy Source Distribution")
plt.show()


# Smartphone Market Share
companies = ["Apple","Samsung","Xiaomi","Oppo","Others"]
market_share = [28,26,20,14,12]

plt.pie(market_share, labels=companies, autopct="%1.1f%%")
plt.title("Technology: Smartphone Market Share")
plt.show()


# Daily Time Distribution
activities = ["Sleep","Work","Study","Exercise","Leisure"]
hours = [32,33,15,5,15]

plt.pie(hours, labels=activities, autopct="%1.1f%%")
plt.title("Lifestyle: Daily Time Distribution")
plt.show()
```
