## Student Exam Performance Analysis using NumPy and Pandas

```python
import pandas as pd
import numpy as np

df = pd.read_csv("StudentsPerformance.csv")

print(df.head())

print(df.info())

print(df.describe())
```

```python
print("Average Math Score:", np.mean(df['math score']))
print("Average Reading Score:", np.mean(df['reading score']))
print("Average Writing Score:", np.mean(df['writing score']))
```

```python
print("Max Math Score:", np.max(df['math score']))
print("Min Math Score:", np.min(df['math score']))
```

```python
gender_avg = df.groupby('gender')[['math score','reading score','writing score']].mean()
print(gender_avg)
```

```python
top_math = df[df['math score'] > 90]
print(top_math)
```

```python
df['total_score'] = df['math score'] + df['reading score'] + df['writing score']
print(df.head())
```

```python
top_students = df.sort_values(by='total_score', ascending=False).head()
print(top_students)
```
