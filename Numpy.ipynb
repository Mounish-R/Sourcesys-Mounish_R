```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([4, 5, 6, 7])

print(a / b)
print(a + 10)
print(a * 3)
print(np.sum(a))
print(np.max(a))
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([4, 5, 6, 7])

print(np.std(a))
print(np.std(b))
```

```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

# axis=0  column-wise
# axis=1  row-wise

print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
```
