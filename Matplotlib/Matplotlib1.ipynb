```python
%matplotlib inline
import matplotlib.pyplot as plt

# Data
x = [1,2,3,4,5]
students = [30,45,50,65,80]
placements = [10,20,30,45,60]

# Create plot
plt.figure(figsize=(8,5))

plt.plot(
    x, students,
    color="purple",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=8
)

plt.plot(
    x, placements,
    color="blue",
    linestyle="-",
    linewidth=3,
    marker="s",
    markersize=8
)

# Annotation
plt.annotate(
    "Highest Placement",
    xy=(5,60),
    xytext=(3,40),
    arrowprops=dict(facecolor="black")
)

# Labels
plt.xlabel("Year")
plt.ylabel("Number of Students")
plt.title("Student Placement Growth")

# Legend
plt.legend(["Total Students","Placed Students"])

# Grid
plt.grid(True)

plt.show()
```
