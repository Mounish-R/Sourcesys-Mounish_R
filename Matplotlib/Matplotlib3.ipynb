## Advanced visualization with subplots

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2, figsize=(10,8))

# Scatter Plot
hours = [1,2,3,4,5]
marks = [35,50,65,40,60]

ax[0,0].scatter(hours, marks, color="blue")
ax[0,0].set_title("Study Hours vs Marks")
ax[0,0].set_xlabel("Hours")
ax[0,0].set_ylabel("Marks")


# Bar Chart
products = ["Laptop","Mobile","Tablet","Camera"]
sales = [120,200,90,60]

ax[0,1].bar(products, sales, color="green")
ax[0,1].set_title("Product Sales")


# Histogram
salaries = [25000,30000,32000,35000,40000,45000,50000]

ax[1,0].hist(salaries, bins=5, color="orange")
ax[1,0].set_title("Salary Distribution")


# Pie Chart
activities = ["Work","Sleep","Study","Leisure"]
hours = [9,7,3,5]

ax[1,1].pie(hours, labels=activities, autopct="%1.1f%%")
ax[1,1].set_title("Daily Time Usage")


plt.tight_layout()
plt.show()
```
