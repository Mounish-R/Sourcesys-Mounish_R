### scatter plot

```python
import matplotlib.pyplot as plt

months = [1,2,3,4,5,6]
temperature = [18,22,26,30,34,38]

plt.figure(figsize=(7,5))

plt.scatter(
    months,
    temperature,
    color="red",
    marker="^",
    s=120,
    edgecolors="black",
    label="Temperature Rise"
)

plt.title("Weather: Monthly Temperature Increase")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")

plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

plt.show()
```

### Bar Chart

```python
subjects = ["Math","Physics","Chemistry","Biology","English"]
scores = [92,85,76,68,60]

plt.figure(figsize=(8,5))

plt.bar(
    subjects,
    scores,
    color=["navy","teal","orange","purple","gray"],
    edgecolor="black"
)

plt.title("Education: Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

### Histogram

```python
rainfall = [20,35,60,85,110,95,70,45,30,25,40,65]

plt.figure(figsize=(7,5))

plt.hist(
    rainfall,
    bins=6,
    color="skyblue",
    edgecolor="black"
)

plt.title("Weather: Rainfall Distribution")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")

plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

### Pie Chart

```python
activities = ["Work","Sleep","Study","Exercise","Leisure"]
hours = [9,7,3,2,3]

explode = [0.1,0,0,0,0]

plt.figure(figsize=(6,6))

plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%",
    colors=["gold","lightblue","lightgreen","orange","pink"],
    explode=explode,
    shadow=True,
    startangle=140
)

plt.title("Daily Time Allocation")

plt.show()
```
